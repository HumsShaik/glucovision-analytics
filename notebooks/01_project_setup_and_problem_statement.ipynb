# HUPA-UCM Diabetes Dataset Analysis

## Project Problem Statement

People with Type 1 Diabetes need to continuously manage glucose levels through insulin, food intake, physical activity, and lifestyle patterns. This project analyzes Continuous Glucose Monitoring data, insulin delivery, carbohydrate intake, heart rate, calories, steps, and patient-level patterns from the HUPA-UCM Diabetes Dataset.

The goal is to identify meaningful glucose patterns, understand which factors are associated with high or low glucose levels, and create a Streamlit dashboard that helps users explore descriptive, prescriptive, and limited predictive insights.

## Business / Healthcare Value

This analysis can help answer questions such as:

- When do glucose levels tend to rise or fall?
- How do insulin, carbohydrate intake, activity, and heart rate relate to glucose?
- Which patients experience more glucose variability?
- Can simple predictive logic help flag possible high or low glucose risk?

## Main Analysis Categories

1. Data Cleaning and Preprocessing
2. Descriptive Analysis
3. Prescriptive / Multivariate Analysis
4. Limited Predictive Analysis
5. Streamlit Dashboard Demo
